```python
#Libraries
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
#settings for plots

#tableau for plots
plt.style.use('tableau-colorblind10')

plt.rcParams["figure.autolayout"] = True

plt.rcParams['font.size'] = 16        # Default font size for text
plt.rcParams['axes.titlesize'] = 16   # Font size for axes titles
plt.rcParams['axes.labelsize'] = 16   # Font size for x and y labels
plt.rcParams['xtick.labelsize'] = 16  # Font size for x tick labels
plt.rcParams['ytick.labelsize'] = 16  # Font size for y tick labels
```

```python
#path to data
path_to_vol = 'placeholder/'
path_to_MFP = 'placeholder/'
```

```python
#sampling a random simulation

files_vol = os.listdir(path_to_vol)
files_MFP = os.listdir(path_to_MFP)

file_vol = files_vol[np.random.randint(0,len(files_vol))]
file_MFP = files_MFP[np.random.randint(0,len(files_MFP))]
```

```python
#load data
data_vol = np.load(path_to_vol + file_vol)
data_MFP = np.load(path_to_MFP + file_MFP)

sim_vol = data_vol['run1']
sim_MFP = data_MFP['run1']
```

```python
#hyperparameters
t = np.linspace(0, 40000*0.005, 40000)
```

```python
#plot volatile phase
plt.figure(figsize=(10,6))

plt.plot(t, sim_vol.T, alpha = 0.8)

plt.xlabel('Time')
plt.ylabel('Population')

plt.text(110, 9, r'$\mu = -3, \sigma = 5, \gamma = -0.5, K = 0.1$')

plt.savefig('results/Ex_Volatile_Phase.png')
```

```python
#plot MFP phase
plt.figure(figsize=(10,6))

plt.plot(t, sim_MFP.T, alpha = 0.8)

plt.xlabel('Time')
plt.ylabel('Population')

plt.ylim(0, 85)

plt.text(110, 80, r'$\mu = -3, \sigma = 5, \gamma = 0.9, K = 0.1$')

plt.savefig('results/Ex_Multiple_Fixed_Points.png')
```
