```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#settings for plots

#tableau for plots
plt.style.use('tableau-colorblind10')

plt.rcParams["figure.autolayout"] = True

plt.rcParams['font.size'] = 16        # Default font size for text
plt.rcParams['axes.titlesize'] = 16   # Font size for axes titles
plt.rcParams['axes.labelsize'] = 16   # Font size for x and y labels
plt.rcParams['xtick.labelsize'] = 16  # Font size for x tick labels
plt.rcParams['ytick.labelsize'] = 16  # Font size for y tick labelsa
```

```python
eta = 0
s = 2
x = np.linspace(-5, 5, 100)
LHS = x
RHS = eta + s*np.tanh(x)
```

```python
plt.figure(figsize=(8, 6))

plt.plot(x, LHS, label=r'$x_{\alpha}$', color = 'blue')
plt.plot(x, RHS, label=r'$s_{\alpha}\tanh(x_{\alpha})+\eta$', color = 'red')

plt.hlines(s+eta, -5, 5, color='black', linestyle='--', alpha = 0.5, label=r'$\pm s_{\alpha}+\eta$')
plt.hlines(-s+eta, -5, 5, color='black', linestyle='--', alpha = 0.5)
plt.xlabel(r'$x_{\alpha}$')
plt.text(-1, 4.5, r'$s_{\alpha} = 2$, $\eta = 0$', fontsize=16)

plt.legend()
plt.savefig('ex_geom_sol_s>1.png', pad_inches=0.1, bbox_inches='tight')
```
